## 自动求导 (Autograd)
PyTorch 以其自动求导（自动微分）功能而闻名。我们可以调用 `backward()` 方法让 `PyTorch` 计算梯度，这些梯度随后会被存储在 `grad` 属性中。

```
# 创建一个示例张量
# requires_grad 参数指示 PyTorch 存储该张量的梯度
x = torch.tensor([2.], requires_grad=True)

# 打印已计算的梯度
# 目前为 None，因为尚未对任何以 x 为自变量的函数求导
pp.pprint(x.grad)
```

```
# 计算 y 关于 x 的梯度
y = x * x * 3 # 3x^2
y.backward()
pp.pprint(x.grad) # d(y)/d(x) = d(3x^2)/d(x) = 6x = 12
```

让我们再次从另一个张量运行反向传播，看看会发生什么。

```
z = x * x * 3 # 3x^2
z.backward()
pp.pprint(x.grad)
```

```
x.grad = None
z = x * x * 3 # 3x^2
z.backward()
# y = x * x * 3
pp.pprint(x.grad)
```

```
z = x * x * 3 # 3x^2
z.backward()
# y = x * x * 3
pp.pprint(x.grad)
```

```
z = x * x * 3 # 3x^2
z.backward()
# y = x * x * 3
pp.pprint(x.grad)
```

我们可以看到，`x.grad` 会被更新为迄今为止计算出的梯度的累加和。当我们在神经网络中运行反向传播时，我们会在进行参数更新之前，把某个特定神经元的所有梯度求和。这正是这里所发生的事情！这也是为什么我们在每次训练迭代中都需要运行 `zero_grad()` 的原因（稍后会详细介绍）。否则，我们的梯度就会在不同的训练迭代之间不断累积，从而导致参数更新出错。

```
# 创建输入数据
input = torch.ones(2,3,4)
# N* H_in -> N*H_out


# 创建一个线性层，将维度为 N,*,H_in 的输入转换为维度为 N,*,H_out 的输出
linear = nn.Linear(4, 2)
linear_output = linear(input)
linear_output
```

### 自定义模块 (Custom Modules)

除了使用预定义的模块外，我们还可以通过继承 `nn.Module` 类来构建我们自己的模块。例如，我们可以利用前面介绍的张量，自己动手构建一个 `nn.Linear`（它也是继承自 `nn.Module`）！我们还可以构建全新的、更复杂的模块，比如自定义的神经网络。您将在稍后的作业中进行这些练习。

要创建一个自定义模块，我们首先要做的是继承 `nn.Module`。然后我们可以在 `__init__` 函数中初始化参数，首先要调用父类的 `__init__` 函数。我们定义的凡是 `nn` 模块对象的所有类属性都会被视为网络参数，它们在训练过程中是可以被学习更新的。普通的张量不是参数，但如果将它们封装在 `nn.Parameter` 类中，就可以转化为网络参数。

所有继承了 `nn.Module` 的类都需要实现一个 `forward(x)` 函数，其中 `x` 是一个输入张量。当我们把数据传递给模块时（例如执行 `model(x)`），该函数就会被调用。

```
class MultilayerPerceptron(nn.Module):

  def __init__(self, input_size, hidden_size):
    # 调用父类的 __init__ 函数
    super(MultilayerPerceptron, self).__init__()

    # 记账（簿记）：保存初始化参数
    self.input_size = input_size
    self.hidden_size = hidden_size

    # 定义我们的模型
    # `self.model` 的命名并没有什么特殊要求。它可以是任意其他名称。
    self.model = nn.Sequential(
        nn.Linear(self.input_size, self.hidden_size),
        nn.ReLU(),
        nn.Linear(self.hidden_size, self.input_size),
        nn.Sigmoid()
    )

  def forward(self, x):
    output = self.model(x)
    return output
```

这是定义同一个类的另一种替代方法。您会发现，我们可以通过在 `__init__` 方法中定义各个单独的层，并在 `forward` 方法中连接它们，从而替代 `nn.Sequential`。

```
class MultilayerPerceptron(nn.Module):

  def __init__(self, input_size, hidden_size):
    # 调用父类的 __init__ 函数
    super(MultilayerPerceptron, self).__init__()

    # 记账（簿记）：保存初始化参数
    self.input_size = input_size
    self.hidden_size = hidden_size

    # 定义我们的网络层
    self.linear = nn.Linear(self.input_size, self.hidden_size)
    self.relu = nn.ReLU()
    self.linear2 = nn.Linear(self.hidden_size, self.input_size)
    self.sigmoid = nn.Sigmoid()

  def forward(self, x):
    linear = self.linear(x)
    relu = self.relu(linear)
    linear2 = self.linear2(relu)
    output = self.sigmoid(linear2)
    return output
```

现在我们已经定义了我们的类，我们可以实例化它并查看它的作用。

```
# 构造一个样本输入
input = torch.randn(2, 5)

# 创建我们的模型
model = MultilayerPerceptron(5, 3)

# 将输入数据传给我们的模型
model(input)
```

我们可以使用 `named_parameters()` 和 `parameters()` 方法来检查我们模型的参数。

```
list(model.named_parameters())
```

## 优化 (Optimization)
我们已经展示了如何使用 `backward()` 函数计算梯度。但仅有梯度不足以让我们的模型进行学习。我们还需要知道如何更新模型的参数。这就是优化器（optimizers）的作用所在。`torch.optim` 模块转换了几个我们可以使用的优化器。一些流行的例子包括 `optim.SGD` 和 `optim.Adam`。在初始化优化器时，我们传入模型参数（可以通过 `model.parameters()` 访问），告诉优化器它将优化哪些值。优化器还具有一个学习率 (`lr`) 参数，用于决定每一步更新的幅度大小。不同的优化器也拥有不同的超参数。

```
import torch.optim as optim
```

在我们有了优化函数之后，我们可以定义我们想要优化的 `loss`（损失函数）。我们既可以自己定义损失，也可以使用 `PyTorch` 中预定义好的损失函数之一，例如 `nn.BCELoss()`。现在让我们把所有内容组合在一起！我们将从创建一些虚拟数据开始。

```
# 创建 y 数据
y = torch.ones(10, 5)

# 向我们的目标 y 添加一些噪声以生成输入 x
# 我们希望模型能预测我们的原始数据，尽管有噪声干扰
x = y + torch.randn_like(y)
x
```

现在，我们可以定义我们的模型、优化器和损失函数。

```
# 实例化模型
model = MultilayerPerceptron(5, 3)

# 定义优化器
adam = optim.Adam(model.parameters(), lr=1e-1)

# 使用预定义的损失函数定义损失
loss_function = nn.MSELoss()

# 计算模型当前的表现
y_pred = model(x)
loss_function(y_pred, y).item()
```

让我们看看能否让模型实现更小的损失。现在我们已经具备了所需的一切，可以设置我们的训练循环了。

```
# 设置 epoch 的数量，这决定了训练迭代的次数
n_epoch = 10

for epoch in range(n_epoch):
  # 将梯度清零
  adam.zero_grad()

  # 获取模型的预测结果
  y_pred = model(x)

  # 获取损失值
  loss = loss_function(y_pred, y)

  # 打印统计信息
  print(f"Epoch {epoch}: traing loss: {loss}")

  # 计算梯度
  loss.backward()

  # 迈出一步以优化权重参数
  adam.step()
```

```
list(model.parameters())
```

您可以看到我们的损失正在降低。现在让我们检查模型的预测结果，看看它们是否接近我们的原始 `y`（全部为 `1`）。

```
# 观察模型在训练数据上的表现
y_pred = model(x)
y_pred
```

```
# 创建测试数据并检查模型在测试数据上的表现
x2 = y + torch.randn_like(y)
y_pred = model(x2)
y_pred
```

太棒了！看来我们的模型几乎完美地学会了从我们传入的 `x` 中过滤掉噪声！

#### 预处理

为了让我们的模型更容易学习，我们通常会向数据应用一些预处理步骤。这在处理文本数据时尤为重要。以下是文本预处理的一些示例：
* **分词 (Tokenization)**：将句子切分为单词。
* **转小写 (Lowercasing)**：将所有字母转换为小写。
* **噪声去除 (Noise removal)**：去除特殊字符（如标点符号）。
* **停用词去除 (Stop words removal)**：去除常用词。

需要哪些预处理步骤取决于手头的任务。例如，尽管在某些任务中去除特殊字符很有用，但对其他任务来说，它们可能很重要（例如，如果我们处理多种语言）。对于我们的任务，我们将单词全部转为小写并进行分词。


```
# 我们将用来生成训练样本的预处理函数
# 我们的函数非常简单：将字母转换为小写，
# 然后对单词进行分词。
def preprocess_sentence(sentence):
  return sentence.lower().split()

# 创建我们的训练集
train_sentences = [preprocess_sentence(sent) for sent in corpus]
train_sentences
```

对于我们拥有的每个训练样本，我们还应该有一个对应的标签。回顾一下，我们模型的目的是确定哪些单词对应于 `LOCATION`（地理位置）。也就是说，我们希望模型对所有不是 `LOCATION` 的单词输出 `0`，对是 `LOCATION` 的单词输出 `1`。

```
# 出现在我们语料库中的地点集合
locations = set(["australia", "ankara", "paris", "stanford", "taiwan", "turkey"])

# 我们的训练标签
train_labels = [[1 if word in locations else 0 for word in sent] for sent in train_sentences]
train_labels
```

#### 将单词转换为嵌入 (Embeddings)

让我们更仔细地看一下我们的训练数据。我们拥有的每个数据点都是一个单词序列。另一方面，我们知道机器学习模型处理的是向量中的数字。我们该如何将单词转换为数字呢？您可能会想到词嵌入（embeddings），您是对的！

想象一下，我们有一个嵌入查找表 `E`，其中每一行对应一个词嵌入向量。也就是说，我们词汇表中的每个单词在这个表中都会有一个对应的嵌入行 `i`。每当我们想要查找一个单词的嵌入时，我们将遵循以下步骤：
1. 在嵌入表中找到该单词对应的索引 `i`：`单词 -> 索引`。
2. 在嵌入表中进行索引并获取嵌入向量：`索引 -> 嵌入向量`。

让我们来看第一步。我们应该将词汇表中的所有单词分配给一个对应的索引。我们可以按如下步骤进行：
1. 找出语料库中所有唯一的单词。
2. 为每个单词分配一个索引。

```
# 找出我们语料库中所有唯一的单词
vocabulary = set(w for s in train_sentences for w in s)
vocabulary
```

`vocabulary` 现在包含了我们语料库中的所有单词。另一方面，在测试期间，我们可能会遇到未包含在词汇表中的单词。如果我们能够找到一种方法来表示这些未知词，由于我们还会查看每个预测值周围的邻近词，模型仍然能够推断它们是否为 `LOCATION`。

我们引入了一个特殊的标记 `<unk>` 来处理词汇表外的单词（未登录词）。如果愿意，我们可以为未知词标记选择其他字符串。这里唯一的的要求是我们的标记应当是唯一的：我们应当只在未知词时使用该标记。我们也将这个特殊标记添加到我们的词汇表中。

```
# 将代表未知词的标记添加到词汇表中
vocabulary.add("<unk>")
```

前面我们提到，我们的任务被称为“字窗口分类 (Word Window Classification)”，因为当模型需要做出预测时，除了给定的目标单词外，还会查看其周围的单词。

例如，让我们以句子 "We always come to Paris" 为例。该句子对应的训练标签是 `0, 0, 0, 0, 1`，因为只有最后一个单词 Paris 是 `LOCATION`。在一次前向传播（即调用 `forward()`）中，我们的模型将试图为一个单词生成正确的标签。假设我们的模型正在试图为 `Paris` 生成正确的标签 `1`。如果我们仅允许模型看到 `Paris`，而看不到其他内容，我们将错失一个重要信息：单词 `to` 经常出现在 `LOCATION` 之前。

字窗口允许我们的模型在进行预测时，考虑每个单词周围的 `+N` 或 `-N` 个单词。在前面关于 `Paris` 的例子中，如果我们的窗口大小为 1，这意味着模型将查看紧接在 `Paris` 之前和之后的单词，也就是 `to`，以及，呃，什么都没有。现在，这引发了另一个问题。`Paris` 位于句子的末尾，因此后面没有其他单词了。请记住，我们在初始化 `PyTorch` 模型时就定义了它们的 input 维度。如果我们将窗口大小设置为 `1`，这意味着我们的模型每次前向传播都要接收 `3` 个单词。我们不能让我们的模型时而期望接收 `2` 个单词，时而接收 `3` 个单词。

解决方案是引入一个特殊的标记，例如 `<pad>`（填充标记），将其添加到句子中，以确保每个单词周围都有一个有效的窗口。与 `<unk>` 标记类似，如果愿意，我们也可以为填充标记选择另一个字符串，只要确保它被用于唯一的特定目的即可。

```
# 将 <pad> 标记添加到词汇表中
vocabulary.add("<pad>")

# 对给定句子进行填充的函数
# 我们在这里引入该函数作为示例
# 我们将在本教程稍后使用它
def pad_window(sentence, window_size, pad_token="<pad>"):
  window = [pad_token] * window_size
  return window + sentence + window

# 展示填充示例
window_size = 2
pad_window(train_sentences[0], window_size=window_size)
```

现在我们的词汇表已经准备就绪，让我们为每一个单词分配一个索引。

```
# 我们只需将词汇表转换为列表形式，以便能够进行索引
# 排序并不是必需的，我们排序是为了展示一个有序的 word_to_ind 字典
# 话虽如此，我们会发现将填充标记的索引设为 0 是很方便的，
# 因为一些 PyTorch 函数（例如我们稍后将介绍的 nn.utils.rnn.pad_sequence）将其用作默认值
ix_to_word = sorted(list(vocabulary))

# 创建一个字典以查找给定单词的索引
word_to_ix = {word: ind for ind, word in enumerate(ix_to_word)}
word_to_ix
```

```
ix_to_word[1]
```

太棒了！我们已经准备好将训练句子转换为对应于每个标记的索引序列。

```
# 给定一个标记形式的句子，返回对应的索引序列
def convert_token_to_indices(sentence, word_to_ix):
  indices = []
  for token in sentence:
    # 检查标记是否在我们的词汇表中。如果在，获取其索引。
    # 如果不在，获取未知标记 <unk> 的索引。
    if token in word_to_ix:
      index = word_to_ix[token]
    else:
      index = word_to_ix["<unk>"]
    indices.append(index)
  return indices

# 同一个函数的更紧凑版本
def _convert_token_to_indices(sentence, word_to_ix):
  return [word_to_ind.get(token, word_to_ix["<unk>"]) for token in sentence]

# 展示一个示例
example_sentence = ["we", "always", "come", "to", "kuwait"]
example_indices = convert_token_to_indices(example_sentence, word_to_ix)
restored_example = [ix_to_word[ind] for ind in example_indices]

print(f"Original sentence is: {example_sentence}")
print(f"Going from words to indices: {example_indices}")
print(f"Going from indices to words: {restored_example}")
```

在上面的示例中，`kuwait` 显示为了 `<unk>`，因为它没有包含在我们的词汇表中。让我们将 `train_sentences` 转换为 `example_padded_indices`。

```
# 将我们的句子转换为索引序列
example_padded_indices = [convert_token_to_indices(s, word_to_ix) for s in train_sentences]
example_padded_indices
```

现在我们已经为词汇表中的每个单词分配了索引，我们可以使用 `PyTorch` 中的 `nn.Embedding` 类来创建一个词嵌入表。它的调用方式如下：`nn.Embedding(num_words, embedding_dimension)`，其中 `num_words` 是我们词汇表中的单词数量，而 `embedding_dimension` 是我们希望拥有的嵌入维度。`nn.Embedding` 并没有什么神奇之处：它只是一个围绕可训练的 `NxE` 维张量的封装类，其中 `N` 是词汇表大小，`E` 是嵌入维度大小。该表最初是随机初始化的，但会随着时间而改变。随着我们训练网络，梯度将一直反向传播到嵌入层，从而更新我们的词嵌入。我们将在模型中初始化用于该模型的嵌入层，这里只是展示一个示例。

```
# 为我们的单词创建一个嵌入表
embedding_dim = 5
embeds = nn.Embedding(len(vocabulary), embedding_dim)

# 打印我们嵌入表中的参数
list(embeds.parameters())
```

要获取我们词汇表中某个单词的词嵌入，我们只需要创建一个查找张量。查找张量只是一个包含我们要查找的索引的张量。`nn.Embedding` 类期望传入一个 Long Tensor 类型的索引张量，因此我们应该相应地创建我们的张量。

```
# 获取单词 Paris 的嵌入向量
index = word_to_ix["paris"]
index_tensor = torch.tensor(index, dtype=torch.long)
paris_embed = embeds(index_tensor)
paris_embed
```

#### 句子批处理 (Batching Sentences)

我们在课堂上已经学习了批处理（batches）。在进行参数更新之前等待整个训练语料库被处理完是代价高昂的。另一方面，在每个训练样本之后就更新参数会导致两次更新之间的损失不够稳定。为了解决这些问题，我们选择在训练完一批（batch）数据后更新参数。这能让我们对全局损失的梯度有更好的估计。在本节中，我们将学习如何使用 `torch.utils.data.DataLoader` 类将数据组织成批次。

我们将按如下方式调用 `DataLoader` 类：`DataLoader(data, batch_size=batch_size, shuffle=True, collate_fn=collate_fn)`。`batch_size` 参数决定了每个批次中的样本数量。在每个 epoch 中，我们将使用 `DataLoader` 迭代所有的批次。默认情况下，批次的顺序是确定的，但我们可以通过将 `shuffle` 参数设置为 `True` 来让 `DataLoader` 打乱批次顺序。通过这种方式，我们能确保不会多次遇到相同的糟糕数据排列批次。

如果提供了 `collate_fn`，`DataLoader` 会将它准备好的批次传递给 `collate_fn`。我们可以编写一个自定义函数传递给 `collate_fn` 参数，以打印关于我们批次的统计信息或进行额外的处理。在我们的例子中，我们将使用 `collate_fn` 来：
1. 对我们的训练句子进行窗口填充。
2. 将训练样本中的单词转换为索引。
3. 对训练样本进行填充，以便所有句子和标签具有相同的长度。同样地，我们也需要对标签进行填充。这会产生一个问题，因为在计算损失时，我们需要知道给定样本中实际的单词数量。我们还将在传递给 `collate_fn` 参数的函数中追踪这个数量。

因为我们的自定义 `collate_fn` 函数版本将需要访问我们的 `word_to_ix` 字典（以便将单词转换为索引），我们将利用 `Python` 中的 `partial` 函数，它能将我们给定的参数传递给包裹的函数。

```
from torch.utils.data import DataLoader
from functools import partial

def custom_collate_fn(batch, window_size, word_to_ix):
  # 将批次拆分为训练样本 (x) 和标签 (y)
  # 我们将 x 和 y 转换为张量，因为 nn.utils.rnn.pad_sequence
  # 方法期望接收张量。这也很有用，因为我们的模型也将
  # 期望接收张量作为输入。
  x, y = zip(*batch)

  # 现在我们需要对训练样本进行窗口填充。我们已经定义过一个
  # 处理窗口填充的函数。我们在这里再次包含它，以便
  # 所有内容都放在同一个地方。
  def pad_window(sentence, window_size, pad_token="<pad>"):
    window = [pad_token] * window_size
    return window + sentence + window

  # 对训练样本进行填充。
  x = [pad_window(s, window_size=window_size) for s in x]

  # 现在我们需要将训练样本中的单词转换为索引。
  # 我们出于与上述相同的原因，复制先前定义好的函数。
  def convert_tokens_to_indices(sentence, word_to_ix):
    return [word_to_ix.get(token, word_to_ix["<unk>"]) for token in sentence]

  # 将训练样本转换为索引。
  x = [convert_tokens_to_indices(s, word_to_ix) for s in x]

  # 我们现在将对样本进行填充，使一个批次中所有样本的长度
  # 保持一致，从而可以进行矩阵运算。
  # 我们将 batch_first 参数设置为 True，以便返回的矩阵将
  # 批次（batch）大小作为第一个维度。
  pad_token_ix = word_to_ix["<pad>"]

  # pad_sequence 函数期望输入为张量，因此我们将 x 转换为张量
  x = [torch.LongTensor(x_i) for x_i in x]
  x_padded = nn.utils.rnn.pad_sequence(x, batch_first=True, padding_value=pad_token_ix)

  # 我们也将对标签进行填充。在此之前，我们将记录标签的
  # 数量，以便知道每个样本中原本存在多少个单词。
  lengths = [len(label) for label in y]
  lenghts = torch.LongTensor(lengths)

  y = [torch.LongTensor(y_i) for y_i in y]
  y_padded = nn.utils.rnn.pad_sequence(y, batch_first=True, padding_value=0)

  # 我们现在准备好返回变量。我们在此返回变量的顺序
  # 将与我们在训练循环中读取它们的顺序相匹配。
  return x_padded, y_padded, lenghts
```

这个函数看起来很长，但实际上没必要这么长。查看下面的替代版本，其中我们删除了额外的函数声明和注释。

```
def _custom_collate_fn(batch, window_size, word_to_ix):
  # 准备数据点
  x, y = zip(*batch)
  x = [pad_window(s, window_size=window_size) for s in x]
  x = [convert_tokens_to_indices(s, word_to_ix) for s in x]

  # 填充 x，使批次中的所有样本大小相同
  pad_token_ix = word_to_ix["<pad>"]
  x = [torch.LongTensor(x_i) for x_i in x]
  x_padded = nn.utils.rnn.pad_sequence(x, batch_first=True, padding_value=pad_token_ix)

  # 填充 y 并记录长度
  lengths = [len(label) for label in y]
  lenghts = torch.LongTensor(lengths)
  y = [torch.LongTensor(y_i) for y_i in y]
  y_padded = nn.utils.rnn.pad_sequence(y, batch_first=True, padding_value=0)

  return x_padded, y_padded, lenghts
```

现在，我们可以 see the `DataLoader` in action.

```
# 要传递给 DataLoader 的参数
data = list(zip(train_sentences, train_labels))
batch_size = 2
shuffle = True
window_size = 2
collate_fn = partial(custom_collate_fn, window_size=window_size, word_to_ix=word_to_ix)

# 实例化 DataLoader
loader = DataLoader(data, batch_size=batch_size, shuffle=shuffle, collate_fn=collate_fn)

# 进行一次循环遍历
counter = 0
for batched_x, batched_y, batched_lengths in loader:
  print(f"Iteration {counter}")
  print("Batched Input:")
  print(batched_x)
  print("Batched Labels:")
  print(batched_y)
  print("Batched Lengths:")
  print(batched_lengths)
  print("")
  counter += 1
```

您在上面看到的批处理输入张量将被传递到我们的模型中。另一方面，我们一开始提到我们的模型将是一个字窗口分类器。以我们目前的输入张量格式化方式，一个数据点包含了一个句子中的所有单词。当我们把这个输入传给我们的模型时，它需要为每个单词创建窗口，针对每个窗口预测中心词是否是 `LOCATION`，最后将这些预测组合在一起并返回。

如果我们事先将数据切分为窗口，就可以避免这个问题。但在本示例中，我们将让我们的模型本身来处理这个格式化过程。

由于我们的 `window_size` 是 `N`，我们希望我们的模型在每 `2N+1` 个标记上进行预测。也就是说，如果我们有一个包含 `9` 个标记的输入，并且 `window_size` 为 `2`，我们希望模型返回 `5` 个预测值。这很合理，因为在我们对两边各自填充 `2` 个标记之前，我们的输入正好也是 `5` 个标记！

我们可以通过使用 for 循环来创建这些窗口，但是有一个更高效的 `PyTorch` 替代方法，即 `unfold(dimension, size, step)` 方法。我们可以使用该方法按如下方式创建我们所需的窗口：

```
# 打印原始张量
print(f"Original Tensor: ")
print(batched_x)
print("")

# 创建大小为 2 * 2 + 1 的数据块 (chunks)
chunk = batched_x.unfold(1, window_size*2 + 1, 1)
print(f"Windows: ")
print(chunk)
```

### 模型

既然我们已经准备好了数据，我们现在就可以开始构建我们的模型了。我们已经学习了如何编写自定义的 `nn.Module` 类。在这里，我们将采取同样的方法，并将迄今为止学到的所有内容结合在一起。

```
class WordWindowClassifier(nn.Module):

  def __init__(self, hyperparameters, vocab_size, pad_ix=0):
    super(WordWindowClassifier, self).__init__()

    """ 实例变量 """
    self.window_size = hyperparameters["window_size"]
    self.embed_dim = hyperparameters["embed_dim"]
    self.hidden_dim = hyperparameters["hidden_dim"]
    self.freeze_embeddings = hyperparameters["freeze_embeddings"]

    """ 嵌入层
    输入一个包含嵌入索引的张量，并返回对应的嵌入向量。
    输出维度为 (number_of_indices * embedding_dim)。

    如果 freeze_embeddings 为 True，则将嵌入层参数设置为不可训练。
    这在我们只想让除嵌入参数之外的其他参数改变时非常有用。

    """
    self.embeds = nn.Embedding(vocab_size, self.embed_dim, padding_idx=pad_ix)
    if self.freeze_embeddings:
      self.embed_layer.weight.requires_grad = False

    """ 隐藏层
    """
    full_window_size = 2 * window_size + 1
    self.hidden_layer = nn.Sequential(
      nn.Linear(full_window_size * self.embed_dim, self.hidden_dim),
      nn.Tanh()
    )

    """ 输出层
    """
    self.output_layer = nn.Linear(self.hidden_dim, 1)

    """ 概率值
    """
    self.probabilities = nn.Sigmoid()

  def forward(self, inputs):
    """
    令 B:= batch_size (批次大小)
        L:= 经过窗口填充的句子长度
        D:= self.embed_dim (嵌入维度)
        S:= self.window_size (窗口大小)
        H:= self.hidden_dim (隐藏层维度)

    inputs: 一个形状为 (B, L) 的 token 索引张量
    """
    B, L = inputs.size()

    """
    重塑形状 (Reshaping)。
    输入一个 (B, L) 的 LongTensor
    输出一个 (B, L~, S) 的 LongTensor
    """
    # 首先，获取输入中每个单词的字窗口 (word windows)。
    token_windows = inputs.unfold(1, 2 * self.window_size + 1, 1)
    _, adjusted_length, _ = token_windows.size()

    # 在代码中（至少在注释里）进行内部张量维度的健全性检查是一个好习惯！
    assert token_windows.size() == (B, adjusted_length, 2 * self.window_size + 1)

    """
    嵌入 (Embedding)。
    输入一个大小为 (B, L~, S) 的 torch.LongTensor
    输出一个 (B, L~, S, D) 的 FloatTensor。
    """
    embedded_windows = self.embeds(token_windows)

    """
    重塑形状 (Reshaping)。
    输入一个 (B, L~, S, D) 的 FloatTensor。
    将其调整为 (B, L~, S*D) 的 FloatTensor。
    参数 -1 表示根据剩余的轴自动推断最后一个维度的大小。
    """
    embedded_windows = embedded_windows.view(B, adjusted_length, -1)

    """
    第 1 层 (Layer 1)。
    输入一个 (B, L~, S*D) 的 FloatTensor。
    将其调整为 (B, L~, H) 的 FloatTensor。
    """
    layer_1 = self.hidden_layer(embedded_windows)

    """
    第 2 层 (Layer 2)
    输入一个 (B, L~, H) 的 FloatTensor。
    将其调整为 (B, L~, 1) 的 FloatTensor。
    """
    output = self.output_layer(layer_1)

    """
    Softmax。
    输入一个未归一化类别得分的 (B, L~, 1) FloatTensor。
    输出一个 (对数) 归一化类别得分的 (B, L~, 1) FloatTensor。
    """
    output = self.probabilities(output)
    output = output.view(B, -1)

    return output
```

### 训练

我们现在准备好把一切组合在一起了。让我们先从准备数据 and 初始化模型开始。然后我们可以初始化优化器并定义我们的损失函数。这一次，我们将定义我们自己的损失函数，而不是像之前那样使用预定义的损失函数。

```
# 准备数据
data = list(zip(train_sentences, train_labels))
batch_size = 2
shuffle = True
window_size = 2
collate_fn = partial(custom_collate_fn, window_size=window_size, word_to_ix=word_to_ix)

# 实例化 DataLoader
loader = DataLoader(data, batch_size=batch_size, shuffle=shuffle, collate_fn=collate_fn)

# 初始化模型
# 将模型的所有超参数放在一个字典中是很有用的做法
model_hyperparameters = {
    "batch_size": 4,
    "window_size": 2,
    "embed_dim": 25,
    "hidden_dim": 25,
    "freeze_embeddings": False,
}

vocab_size = len(word_to_ix)
model = WordWindowClassifier(model_hyperparameters, vocab_size)

# 定义优化器
learning_rate = 0.01
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# 定义损失函数，计算得到二元交叉熵损失
def loss_function(batch_outputs, batch_labels, batch_lengths):
    # 计算整个批次的损失
    bceloss = nn.BCELoss()
    loss = bceloss(batch_outputs, batch_labels.float())

    # 重新缩放损失值。记住我们曾使用 lengths 来存储
    # 每个训练样本中的实际单词数量
    loss = loss / batch_lengths.sum().float()

    return loss
```

与我们之前的示例不同，这一次我们不再是在每个 epoch 中一次性将所有训练数据传递给模型，而是利用批次。因此，在每一次训练 epoch 迭代中，我们也会遍历各个批次。

```
# 每个 epoch 中都会被调用的函数
def train_epoch(loss_function, optimizer, model, loader):

  # 追踪该批次的总损失
  total_loss = 0
  for batch_inputs, batch_labels, batch_lengths in loader:
    # 清空梯度
    optimizer.zero_grad()
    # 进行前向传播
    outputs = model.forward(batch_inputs)
    # 计算批次损失
    loss = loss_function(outputs, batch_labels, batch_lengths)
    # 计算梯度
    loss.backward()
    # 更新参数
    optimizer.step()
    total_loss += loss.item()

  return total_loss


# 包含我们主训练循环的函数
def train(loss_function, optimizer, model, loader, num_epochs=10000):

  # 遍历每个 epoch 并调用我们的 train_epoch 函数
  for epoch in range(num_epochs):
    epoch_loss = train_epoch(loss_function, optimizer, model, loader)
    if epoch % 100 == 0: print(epoch_loss)
```

让我们开始训练吧！

```
num_epochs = 1000
train(loss_function, optimizer, model, loader, num_epochs=num_epochs)
```
