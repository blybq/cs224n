Python 具有几种用于存储和操作数据的内置类型：list (列表)、tuple (元组)、dict (字典)。这里是关于这些类型（以及许多其他类型）的官方 Python 文档：https://docs.python.org/3/library/stdtypes.html。

### 列表

列表是可变数组。让我们看看它们是如何工作的。

```python
names = ["Zach", "Jay"]
```

```python
# 通过索引访问列表元素
print(names[0])
```

```python
# 向列表追加元素（追加到列表末尾）
names.append("Richard")
print(names)
```

```python
# 获取列表长度
print(len(names))
```

```python
# 拼接两个列表
# += 运算符是 list1 = list1 + list2 的简写（也可以用于 -, *, / 以及其他类型的变量）
names += ["Abi", "Kevin"]
print(names)
```

```python
# 创建空列表的两种方式
more_names = []
more_names = list()
```

```python
# 创建包含不同数据类型的列表，这在 Python 中是允许的
stuff = [1, ["hi", "bye"], -0.12, None]
print(stuff)
```

列表切片是访问列表中一部分元素的有用方法。

```python
numbers = [0, 1, 2, 3, 4, 5, 6]

# 切片范围从起始索引（包含）到结束索引（不包含）
print(numbers[0:3])
```

```python
# 未指定起始索引时，默认从列表开头开始
# 未指定结束索引时，默认到列表末尾结束
print(numbers[:3])
print(numbers[5:])
```

```python
# : 获取某一维度上所有元素的切片，在处理 NumPy 数组时非常有用
print(numbers[:])
```

```python
# 负索引会绕回，从列表末尾开始倒数计数
print(numbers[-1])
print(numbers[-3:])
print(numbers[3:-2])
```

### 元组

元组是不可变数组。让我们看看它们是如何工作的。

```python
# 元组使用圆括号，列表使用方括号
names = ("Zach", "Jay")
```

```python
# 访问元素和获取长度的语法与列表相同
print(names[0])
print(len(names))
```

```python
# 但与列表不同，元组不支持元素重新赋值
names[0] = "Richard"
```

```python
# 创建空元组
empty = tuple()
print(empty)

# 创建仅包含单个元素的元组，逗号非常重要
single = (10,)
print(single)
```

## 字典

字典是哈希映射（Hash Map）。让我们看看它们是如何工作的。

```python
# 创建空字典的两种方式
phonebook = {}
phonebook = dict()
```

```python
# 创建包含一个元素的字典 
phonebook = {"Zach": "12-37"}
# 添加另一个元素
phonebook["Jay"] = "34-23"
```

```python
# 检查字典中是否存在某个键
print("Zach" in phonebook)
print("Kevin" in phonebook)
```

```python
# 获取键对应的存储值
print(phonebook["Jay"])
```

```python
# 删除一个元素
del phonebook["Zach"]
print(phonebook)
```

## 循环

```python
# 基础 for 循环
for i in range(5):
    print(i)
```

```python
# 遍历列表
names = ["Zach", "Jay", "Richard"]
for name in names:
    print(name)
```

```python
# 遍历列表的索引和值
# 方法 1
for i in range(len(names)):
    print(i, names[i])
    
print("---")

# 方法 2
for i, name in enumerate(names):
    print(i, name)
```

```python
# 遍历字典
phonebook = {"Zach": "12-37", "Jay": "34-23"}

# 遍历键
for name in phonebook:
    print(name)

print("---")

# 遍历值
for number in phonebook.values():
    print(number)

print("---")

# 遍历键和值
for name, number in phonebook.items():
    print(name, number)
```

接下来，让我们看一些矩阵操作。首先是按元素相乘（哈达玛积，Hadamard product）。

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[3, 3], [3, 3]])
print(A)
print(B)
print("---")
print(A * B)
```

我们可以使用 np.matmul 或 @ 进行矩阵乘法。

```python
# 矩阵乘法的一种方法
print(np.matmul(A, B))

# 矩阵乘法的另一种方法
print(A @ B)
```

我们可以使用 np.dot 计算点积（内积）或矩阵与向量的乘积。

```python
u = np.array([1, 2, 3])
v = np.array([1, 10, 100])

print(np.dot(u, v))

# 也可以直接在 numpy 数组上调用 numpy 操作，这对于链式调用多个操作很有用
print(u.dot(v))
```

```python
W = np.array([[1, 2], [3, 4], [5, 6]])
print(v.shape)
print(W.shape)

# 这行得通。
print(np.dot(v, W))
print(np.dot(v, W).shape)
```

### 索引

对 numpy 数组进行切片/索引是 Python 切片（列表）概念在 N 维空间上的扩展。

```python
x = np.random.random((3, 4))

# 选择 x 中的所有元素
print(x[:])
```

### 广播

术语“广播”描述了 NumPy 在算术运算中如何处理具有不同形状的数组。

**通用广播规则**

在对两个数组进行操作时，NumPy 会按元素比较它们的形状。它从最末尾（即最右侧）的维度开始，向左依次比较。两个维度在以下情况下是兼容的：
- 它们相等，或者
- 其中一个是 1（在这种情况下，该轴上的元素会沿该维度进行复制）

更多细节详见：https://numpy.org/doc/stable/user/basics.broadcasting.html

```python
x = np.random.random((3, 4))

y = np.random.random((3, 1))
z = np.random.random((1, 4))

# 在这个例子中，y 和 z 被广播以匹配 x 的形状。
# y 沿着维度 1 被广播。
s = x + y
# z 沿着维度 0 被广播。
p = x * z
```

```python
print(x.shape)
print()
print(y.shape)
print(s.shape)
```

```python
print(x.shape)
print()
print(s.shape)
print(p.shape)
```

```python
a = np.zeros((3, 3))
b = np.array([[1, 2, 3]])
print(a)
print()
print(a+b)
```

### 高效的 NumPy 代码

处理 numpy 数组时，应不惜一切代价避免对索引/轴进行显式的 for 循环。for 循环会极大地降低代码速度（~10-100 倍）。

我们可以使用 %%timeit 魔法命令来计算代码运行时间。让我们对比一下显式 for 循环与使用 numpy 操作的性能。

```python
%%timeit
x = np.random.rand(1000, 1000)
for i in range(100, 1000):
    for j in range(x.shape[1]): 
        x[i, j] += 5
```

```python
%%timeit
x = np.random.rand(1000, 1000)
x[np.arange(100,1000), :] += 5 
```
